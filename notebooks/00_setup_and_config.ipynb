```python
# ============================================================
# Notebook : 00_setup_and_config.ipynb
# Project  : Causal-Guided Multimodal Fusion for Elderly Risk Detection
# Purpose  : Global paths, reproducibility seeds, and the single shared
#            figure style used by every notebook in this project.
#            (grayscale seaborn / 600 dpi / save both .png and .pdf /
#             no in-figure captions)
# Note     : Anonymous repository for double-blind review.
# ============================================================

import os, sys, random, platform
from pathlib import Path

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns

# ── Reproducibility ──────────────────────────────────────────
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)
try:
    import torch
    torch.manual_seed(SEED)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)
    TORCH_OK = True
except Exception:
    TORCH_OK = False

# ── Project paths ────────────────────────────────────────────
# Notebooks live in notebooks/ ; project root is one level up.
ROOT      = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR  = ROOT / "data"
PROC_DIR  = DATA_DIR / "processed"
FIG_DIR   = ROOT / "results" / "figures"
TAB_DIR   = ROOT / "results" / "tables"
for d in (PROC_DIR, FIG_DIR, TAB_DIR):
    d.mkdir(parents=True, exist_ok=True)

# ── Global figure style (grayscale, publication) ─────────────
sns.set_theme(style="whitegrid", context="paper")
sns.set_palette(sns.color_palette("Greys", n_colors=8))
mpl.rcParams.update({
    "figure.dpi":        150,          # on-screen
    "savefig.dpi":       600,          # on-disk (requirement)
    "savefig.bbox":      "tight",
    "font.family":       "DejaVu Sans",
    "axes.unicode_minus": False,
    "image.cmap":        "Greys",      # grayscale images / heatmaps
    "axes.grid":         True,
    "grid.alpha":        0.3,
})

GRAY = {"dark": "#222222", "mid": "#666666", "light": "#AAAAAA", "line": "#000000"}

def savefig(fig, name: str):
    """Save a figure to results/figures as BOTH .png and .pdf at 600 dpi.
    No caption is added (captions belong in the manuscript)."""
    FIG_DIR.mkdir(parents=True, exist_ok=True)
    for ext in ("png", "pdf"):
        fig.savefig(FIG_DIR / f"{name}.{ext}", dpi=600, bbox_inches="tight")
    plt.close(fig)

def savetable(df, name: str):
    """Save a result table to results/tables as .csv."""
    TAB_DIR.mkdir(parents=True, exist_ok=True)
    df.to_csv(TAB_DIR / f"{name}.csv", index=False)

# ── Report ───────────────────────────────────────────────────
print("Environment")
print(f"  Python   : {platform.python_version()}")
print(f"  NumPy    : {np.__version__}")
print(f"  seaborn  : {sns.__version__}")
print(f"  torch    : {'available' if TORCH_OK else 'NOT installed'}")
print("Paths")
print(f"  ROOT     : {ROOT}")
print(f"  DATA_DIR : {DATA_DIR}")
print(f"  FIG_DIR  : {FIG_DIR}")
print(f"  TAB_DIR  : {TAB_DIR}")

# ── Smoke test: verify the figure style actually writes png+pdf ──
_fig, _ax = plt.subplots(figsize=(4, 3))
sns.lineplot(x=np.arange(10), y=np.sin(np.arange(10) / 2), color=GRAY["line"], ax=_ax)
_ax.set_xlabel("x"); _ax.set_ylabel("value")
savefig(_fig, "00_style_smoketest")
print("\nStyle smoke-test figure written:")
for ext in ("png", "pdf"):
    p = FIG_DIR / f"00_style_smoketest.{ext}"
    print(f"  {p}  ({p.stat().st_size/1024:.1f} KB)" if p.exists() else f"  MISSING {p}")
print("\n00_setup_and_config complete.")
```
